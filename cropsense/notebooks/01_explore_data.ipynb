# CropSense data check

Count images under `data/processed` before training.

```python
from pathlib import Path
from collections import Counter

root = Path("../data/processed")
counts = Counter()
for split in ("train", "val"):
    for folder in (root / split).glob("*"):
        if folder.is_dir():
            n = len(list(folder.glob("*.*")))
            counts[f"{split}/{folder.name}"] = n
counts
```
